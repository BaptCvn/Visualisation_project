```python
import pandas as pd
import bar_chart_race as bcr
from IPython.display import Video
```

```python
# on importe le dataset
data = pd.read_csv('Names_hints\dpt2020.csv', sep=';')
```

```python
# on retire les colonnes inutiles et les valeurs manquantes
data = data.drop(columns=["dpt"])
data = data[data["preusuel"] != "_PRENOMS_RARES"]
data = data[data["annais"] != "XXXX"]
```

```python
data.head()
```

```python
grouped = data.groupby(['annais', 'preusuel'])['nombre'].sum().reset_index()

pivot = grouped.pivot(index='annais', columns='preusuel', values='nombre').fillna(0)

top_prenoms = pivot.sum().sort_values(ascending=False).head(100).index
pivot = pivot[top_prenoms]

pivot.index = pivot.index.astype(int)

# Créer la vidéo avec bar_chart_race
bcr.bar_chart_race(
    df=pivot,
    filename='top_prenoms.mp4',
    orientation='h',
    sort='desc',
    n_bars=10,
    fixed_order=False,
    fixed_max=True,
    steps_per_period=20,
    interpolate_period=True,
    period_length=500,
    period_fmt="{x:.0f}",
    title='Top 10 des prénoms les plus donnés par année en France',
    bar_size=0.95,
    cmap='dark12',
    filter_column_colors=True,
)
```
